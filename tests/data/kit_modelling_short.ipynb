```python
import ai_learning_kit as kit
kit.start("modelling")
T = kit.tools()

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

```python
penguins = T.penguins().dropna().reset_index(drop=True)
train_val, test = train_test_split(penguins, test_size=0.2, random_state=T.SEED, stratify=penguins["species"])
train, val = train_test_split(train_val, test_size=0.25, random_state=T.SEED, stratify=train_val["species"])
print(len(train), len(val), len(test))
```

```python
line = LinearRegression().fit(train[["flipper_length_mm"]], train["body_mass_g"])
slope, intercept = T.line_by_hand(train["flipper_length_mm"], train["body_mass_g"], show=False)
print(round(line.coef_[0], 1), round(slope, 1))
```

```python
T.plot_line(train["flipper_length_mm"], train["body_mass_g"], slope, intercept)
plt.show()
```

```python
measures = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
scaler = StandardScaler().fit(train[measures])
knn = KNeighborsClassifier(n_neighbors=5).fit(scaler.transform(train[measures]), train["species"])
print(round(knn.score(scaler.transform(val[measures]), val["species"]), 2))
```

```python
T.check("E4", 0.875, baseline=0.5625)
```
