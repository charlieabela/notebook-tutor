```python
import ai_learning_kit as kit
kit.start("exploring-data")
T = kit.tools()

import pandas as pd
import matplotlib.pyplot as plt
```

```python
penguins = T.penguins().dropna(subset=["bill_length_mm"])
print(len(penguins), round(penguins["body_mass_g"].median()))
```

```python
r = penguins["flipper_length_mm"].corr(penguins["body_mass_g"], method="spearman")
print(round(r, 2))
```

```python
T.corr_heatmap(penguins[["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]])
```
